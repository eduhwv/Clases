Palabras Reservadas
Las palabras reservadas (keywords) corresponden a los nombres de las declaraciones que el interprete de Python incluye por defecto, No se deben utilizar para asignar nombres a otros objetos.

Ejemplo:

El siguiente comando desplegara la lista de palabras reservadas:

```python
help("keywords")
```

El espacio de nombres (namespace)
El espacio de nombres (namespace) contiene una elacion de objetos existentes en la memoria del sistema y los nombres a los que estan ligados.

Python es un lenguaje de alto nivel en el que todos sus elementos son objetos incluyendo los tipos de datos basicos. Si un objeto no esta ligado al menos a un nombre, dicho objeto es desechado por el interprete. Un objeto puede tener más de un nombre.

El operador de asignación =
Para asignar a un nombre con un objeto, se utiliza el operador de asignación "=" con la siguiente sintaxis.

=

```python
saludo ="Hola"
print(saludo)
```

```python
numero = 9
print(numero)
numero2 =9+1
print(numero2)
numero2 =numero+1
print(numero2)
numero3=2
print(numero3)
numero4 = numero3+1
print(numero4)
numero5 = 45.9
print(numero5)
Aprendio =True
Reprobo =False
print(Aprendio)
print(Reprobo)
```

# Ejemplo 1 Asignación de variables
En Python podemos asignar diferentes valores a los nombres de las variables utilizando el operador =. Después podemos mostrar su contenido mediante la función print().

```python
nombre = "Eduardo"
edad = 18
estatura = 1.68
estudia = True

print(nombre)
print(edad)
print(estatura)
print(estudia)
```

# Ejemplo 2. Modificación de variables
Una variable puede cambiar su valor durante la ejecución del programa. Para hacerlo, utilizamos nuevamente el operador de asignación = y podemos realizar operaciones matemáticas con los valores anteriores.

```python
numero = 45
print(numero)

numero = numero + 10
print(numero)

numero = numero * 6
print(numero)

resultado = numero - 15
print(resultado)
```
